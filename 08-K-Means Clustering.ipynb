# Import Libraries

```python
from sklearn.datasets import make_blobs
from sklearn.cluster import KMeans
import matplotlib.pyplot as plt
from sklearn.metrics import adjusted_rand_score, silhouette_score
```

# Create sample data

```python
X, y = make_blobs(n_samples=600, centers=3, cluster_std=0.6)
```

```python
X
```

```python
y
```

# Split training and Testitng Data

```python
from sklearn.model_selection import train_test_split
```

```python
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size = 0.3, random_state = 0)
```

# Create and train the K-Means model

```python
kmeans = KMeans(n_clusters= 3)
kmeans.fit(X_train)
```

# Prediction

```python
y_pred = kmeans.predict(X_test)
```

```python
print("Pridiction: ", y_pred)
```

```python
print("Actual y_test: ", y_test)
```

# Evalution

```python
adjusted_rand_score(y_test, y_pred)
```

```python
silhouette_score(X, kmeans.fit_predict(X))
```

# Visualizaton

```python
# Plot results
plt.scatter(X[:, 0], X[:, 1], c=kmeans.labels_, cmap='rainbow')
plt.scatter(kmeans.cluster_centers_[:, 0], kmeans.cluster_centers_[:, 1], s=100, c='black', marker='X')
plt.title("K-Means Clustering")
plt.show()
```
