# Activation Functions Visualised

```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
# 1. Define all activation functions (scalar-friendly)
def step(x):
    return np.where(x >= 0, 1, 0)

def sigmoid(x):
    return 1 / (1 + np.exp(-x))

def tanh(x):
    return np.tanh(x)

def relu(x):
    return np.maximum(0, x)

def leaky_relu(x, alpha=0.1):
    return np.where(x >= 0, x, alpha * x)
```

```python
# 2. Create the input range
x = np.linspace(-5, 5, 200)              
```

```python
step(x)
```

```python
sigmoid(x)
```

```python
tanh(x)
```

```python
relu(x)
```

```python
leaky_relu(x)
```

```python
# 3. Set up the figure (2 rows, 3 columns)
fig, axes = plt.subplots(2, 3, figsize=(15, 8))

# ---- Plot 1: Step ----
axes[0, 0].plot(x, step(x), color='black', linewidth=2)
axes[0, 0].set_title('Step Activation')
axes[0, 0].grid(True, alpha=0.3)
axes[0, 0].set_ylim(-0.1, 1.1)

# ---- Plot 2: Sigmoid ----
axes[0, 1].plot(x, sigmoid(x), color='blue', linewidth=2)
axes[0, 1].set_title('Sigmoid Activation')
axes[0, 1].grid(True, alpha=0.3)
axes[0, 1].set_ylim(-0.1, 1.1)

# ---- Plot 3: Tanh ----
axes[0, 2].plot(x, tanh(x), color='green', linewidth=2)
axes[0, 2].set_title('Tanh Activation')
axes[0, 2].grid(True, alpha=0.3)
axes[0, 2].set_ylim(-1.1, 1.1)

# ---- Plot 4: ReLU ----
axes[1, 0].plot(x, relu(x), color='red', linewidth=2)
axes[1, 0].set_title('ReLU Activation')
axes[1, 0].grid(True, alpha=0.3)
axes[1, 0].set_ylim(-0.5, 5)

# ---- Plot 5: Leaky ReLU ----
axes[1, 1].plot(x, leaky_relu(x), color='orange', linewidth=2)
axes[1, 1].set_title('Leaky ReLU (α=0.1)')
axes[1, 1].grid(True, alpha=0.3)
axes[1, 1].set_ylim(-0.5, 5)

# ---- Plot 6: Softmax (Special 3-class visualization) ----
# We treat the input as a 3-element vector: [x, 0, -x]
# As x varies, we compute the softmax over the 3 classes.
logits = np.array([x, np.zeros_like(x), -x])   # shape: (3, 200)
exp_logits = np.exp(logits)
softmax_outputs = exp_logits / np.sum(exp_logits, axis=0, keepdims=True)

# Plot each class probability
axes[1, 2].plot(x, softmax_outputs[0, :], label='Class 1 (x)', linewidth=2)
axes[1, 2].plot(x, softmax_outputs[1, :], label='Class 2 (0)', linewidth=2)
axes[1, 2].plot(x, softmax_outputs[2, :], label='Class 3 (-x)', linewidth=2)
axes[1, 2].set_title('Softmax (logits = [x, 0, -x])')
axes[1, 2].legend()
axes[1, 2].grid(True, alpha=0.3)
axes[1, 2].set_ylim(-0.05, 1.05)

# Adjust layout and show
plt.tight_layout()

plt.savefig('activation_function_plot.png')
plt.show()
```

# Single Neuron (Logistic Regression)

```python
# OR gate data
X = np.array([[0, 0],
              [0, 1],
              [1, 0],
              [1, 1]])
y = np.array([0, 1, 1, 1]).reshape(-1, 1)

# Initialize parameters
np.random.seed(42)
W = np.random.randn(2, 1) * 0.01
b = np.zeros((1, 1))

# Hyperparameters
lr = 0.5
epochs = 2000

# Training loop
for epoch in range(epochs):
    # Forward pass
    Z = np.dot(X, W) + b
    A = 1 / (1 + np.exp(-Z))          # sigmoid
    
    # Compute loss (binary cross-entropy)
    loss = -np.mean(y * np.log(A) + (1 - y) * np.log(1 - A))
    
    # Backward pass (gradients)
    dZ = A - y
    dW = np.dot(X.T, dZ) / len(y)
    db = np.sum(dZ) / len(y)
    
    # Update parameters
    W -= lr * dW
    b -= lr * db
    
    if epoch % 500 == 0:
        print(f"Epoch {epoch}, Loss: {loss:.4f}")

# Test
predictions = (A > 0.5).astype(int)
print("\nPredictions:", predictions.flatten())
print("Actual:     ", y.flatten())
```

# Neural Network from Scratch (XOR Problem)

```python
# XOR data
X = np.array([[0, 0],
              [0, 1],
              [1, 0],
              [1, 1]])
y = np.array([0, 1, 1, 0]).reshape(-1, 1)

# Network architecture
input_size = 2
hidden_size = 4   # enough to learn XOR
output_size = 1

# Initialize weights and biases
np.random.seed(42)
W1 = np.random.randn(input_size, hidden_size) * 0.01
b1 = np.zeros((1, hidden_size))
W2 = np.random.randn(hidden_size, output_size) * 0.01
b2 = np.zeros((1, output_size))

# Activation functions (and their derivatives)
def sigmoid(x):
    return 1 / (1 + np.exp(-x))

def sigmoid_derivative(x):
    return x * (1 - x)   # x is already sigmoid output

# Hyperparameters
lr = 0.5
epochs = 10000

loss_history = []

for epoch in range(epochs):
    # ---------- Forward pass ----------
    Z1 = np.dot(X, W1) + b1
    A1 = sigmoid(Z1)
    Z2 = np.dot(A1, W2) + b2
    A2 = sigmoid(Z2)   # final output
    
    # Loss (binary cross-entropy)
    loss = -np.mean(y * np.log(A2) + (1 - y) * np.log(1 - A2))
    loss_history.append(loss)
    
    # ---------- Backward pass ----------
    # Output layer gradients
    dZ2 = A2 - y
    dW2 = np.dot(A1.T, dZ2) / len(y)
    db2 = np.sum(dZ2, axis=0, keepdims=True) / len(y)
    
    # Hidden layer gradients
    dA1 = np.dot(dZ2, W2.T)
    dZ1 = dA1 * sigmoid_derivative(A1)
    dW1 = np.dot(X.T, dZ1) / len(y)
    db1 = np.sum(dZ1, axis=0, keepdims=True) / len(y)
    
    # Update parameters
    W2 -= lr * dW2
    b2 -= lr * db2
    W1 -= lr * dW1
    b1 -= lr * db1
    
    if epoch % 2000 == 0:
        print(f"Epoch {epoch}, Loss: {loss:.4f}")

# Test the trained model
def predict(x):
    Z1 = np.dot(x, W1) + b1
    A1 = sigmoid(Z1)
    Z2 = np.dot(A1, W2) + b2
    A2 = sigmoid(Z2)
    return (A2 > 0.5).astype(int)

print("\nPredictions:")
for i, inp in enumerate(X):
    pred = predict(inp.reshape(1, -1))
    print(f"{inp} -> {pred[0][0]} (expected {y[i][0]})")

# Plot loss curve
plt.plot(loss_history)
plt.title('Training Loss')
plt.xlabel('Epoch')
plt.ylabel('Binary Cross-Entropy')
plt.grid(True)
plt.show()
```

# Quick Practical Demo with Keras (MNIST)

```python
import tensorflow as tf
from tensorflow.keras import layers, models
from tensorflow.keras.datasets import mnist
import matplotlib.pyplot as plt
```

```python
# Load and preprocess MNIST
(x_train, y_train), (x_test, y_test) = mnist.load_data()
```

```python
x_train.shape
```

```python
x_test.shape
```

```python
x_train = x_train.reshape(-1, 28*28) / 255.0
x_test = x_test.reshape(-1, 28*28) / 255.0
```

```python
x_train.shape
```

```python
x_test.shape
```

```python
# Build a simple feedforward network
model = models.Sequential([
    layers.Dense(128, activation='relu', input_shape=(784,)),
    layers.Dense(64, activation='relu'),
    layers.Dense(10, activation='softmax')
])

model.compile(optimizer='adam',
              loss='sparse_categorical_crossentropy',
              metrics=['accuracy'])
```

```python
# Train
history = model.fit(x_train, y_train, epochs=5, batch_size=32,
                    validation_data=(x_test, y_test))
```

```python
# Evaluate
test_loss, test_acc = model.evaluate(x_test, y_test, verbose=0)
print(f"\nTest accuracy: {test_acc:.4f}")
```

```python
# Plot training history
plt.figure(figsize=(12, 4))
plt.subplot(1, 2, 1)
plt.plot(history.history['accuracy'], label='Train acc')
plt.plot(history.history['val_accuracy'], label='Val acc')
plt.title('Accuracy'); plt.legend(); plt.grid(True)

plt.subplot(1, 2, 2)
plt.plot(history.history['loss'], label='Train loss')
plt.plot(history.history['val_loss'], label='Val loss')
plt.title('Loss'); plt.legend(); plt.grid(True)
plt.show()
```
