## Here’s a simple PCA example using scikit-learn. 
## It reduces the Iris dataset’s four measurements to two principal components, then reports how much variance each component explains.

```python
from sklearn.datasets import load_iris
from sklearn.preprocessing import StandardScaler
from sklearn.decomposition import PCA
```

```python
# Load the data (four measurements per flower)
iris = load_iris()
X = iris.data
```

```python
# PCA is sensitive to scale, so standardize the features first
X_scaled = StandardScaler().fit_transform(X)
```

```python
# Reduce 4 features to 2 principal components
pca = PCA(n_components=2)
X_pca = pca.fit_transform(X_scaled)
```

```python
print("Original shape:", X.shape)
print("Reduced shape:", X_pca.shape)
print("Variance explained by each component:",
      pca.explained_variance_ratio_)
print("Total variance explained:",
      pca.explained_variance_ratio_.sum())
print("First 5 transformed rows:\n", X_pca[:5])
```
